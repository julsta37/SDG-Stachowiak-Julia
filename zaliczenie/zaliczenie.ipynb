```python
import pandas as pd
import geopandas as gpd
import os
```

```python
#Stworzenie DataFrame
data = {
    "nazwa": [],
    "rodzaj geometrii": [],
    "ilość obiektów": [],
    "średnia powierzchnia": [],
    "średnia długość": [],
    "średnia złożoność linii": [],
    "średnie wypełnienie przestrzeni": []
}

df = pd.DataFrame(data)
```

```python
#Wczytanie warstw z BDOT - powiat gołdapski

path = "C:/Users/julia/Desktop/SDG/SDG-Stachowiak-Julia/zaliczenie/dane/"
warstwy = os.listdir(path)

dane = []

for warstwa in warstwy:
    if warstwa.endswith(".shp"):
        gdf = gpd.read_file(os.path.join(path, warstwa))
        
        dane.append({
            "nazwa": warstwa,
            "rodzaj geometrii": gdf.geom_type.unique()[0],
            "ilość obiektów": len(gdf),
            "średnia powierzchnia": gdf.area.mean() if gdf.geom_type.unique() == "Polygon" else "Nie dotyczy",
            "średnia długość": gdf.length.mean() if gdf.geom_type.unique() == "LineString" else "Nie dotyczy",
            "średnia złożoność linii": (gdf.geometry.apply(lambda geom: len(geom.coords)).mean() / gdf.length.mean()) 
                                        if gdf.geom_type.unique() == "LineString" else "Nie dotyczy",
            "średnie wypełnienie przestrzeni": (gdf.area.mean() / gdf.envelope.area.mean()) if gdf.geom_type.unique() == "Polygon" else "Nie dotyczy"
        })

df = pd.DataFrame(dane)
print(df)
```
