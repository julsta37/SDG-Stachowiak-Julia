```python
#Dla warstwy z pasami ukształtowania policz ich powierzchnie w m2, korzystając z układu epsg:2180
import geopandas as gpd

path_to_files = "C:/Users/julia/Desktop/SDG/cwiczenia/dane/"

zones = gpd.read_file(path_to_files+"pl_pasy_ukształtowania.geojson")

print(zones.crs.name)

zones2 = zones.to_crs("EPSG:2180")
print(zones2.crs.name)

zones2_area = zones2.area
print(zones2_area)
```

```python
#Stwórz obiekt Bounding Box dla warstwy z miastami wojewódzkimi, a następnie oblicz jego obwód

cities = gpd.read_file(path_to_files + "pl_miasta_wojewódzkie.geojson")

bbox = cities.to_crs("EPSG:2180").dissolve().envelope
obwod = bbox.length.values[0]
print(f"Obwód bbox wynosi: {obwod / 1000:.2f} km")
```

```python
#Dla warstwy z rzekami stwórz nową kolumnę w której wyliczysz liczbę linii (LineString) znajdujących się wewnątrz. Analogicznie stwórz następnie drugą kolumnę w której wyliczysz łączną liczbę punktów składających się na geometrię

rivers = gpd.read_file(path_to_files+"pl_rzeki.geojson")

rivers['liczba_linii'] = rivers.geometry.apply(
    lambda g: len(g.geoms) if g.geom_type == 'MultiLineString'
    else (1 if g.geom_type == 'LineString' else 0)
)

rivers['liczba_linii']

def count_vertices(geom):
    if geom is None or geom.is_empty:
        return 0
    elif geom.geom_type == 'LineString':
        return len(geom.coords)
    elif geom.geom_type == 'MultiLineString':
        return sum(len(line.coords) for line in geom.geoms)
    return 0

rivers['liczba_punktow'] = rivers.geometry.apply(count_vertices)

rivers['liczba_punktow']
```
